# 04 · Checklist baseline

The high-risk checklist is the baseline every model must beat. Flags are defined in
contracts/cohort_config.yaml.


```python
import pandas as pd

# KE squad: COUNTRY = 'ke'   |   TZ squad: COUNTRY = 'tz'
COUNTRY = 'ke'
COHORT  = 'perinatal'   # perinatal | perinatal_anc | lbw

df = pd.read_csv(f'../../data/processed/{COUNTRY}_{COHORT}.csv')
print(df.shape)
df.head()
```

## Build the checklist flag
age < 18 or ≥ 35 · prior_live_births ≥ 5 · preceding_interval_m < 24 · multiple · prior_stillbirth


```python
flag = ((df['maternal_age']<18)|(df['maternal_age']>=35)
        |(df['prior_live_births']>=5)
        |(df['preceding_interval_m']<24)
        |(df['multiple']==1)
        |(df['prior_stillbirth']==1))
flag.mean()   # share of pregnancies flagged
```

## Score it
- how many perinatal deaths does the checklist catch, and at what flag rate?
- this flag rate is what the model is compared against (same share flagged → events caught)


```python
print('events caught:', df.loc[flag,'perinatal_death'].sum(), 'of', df['perinatal_death'].sum())
```
