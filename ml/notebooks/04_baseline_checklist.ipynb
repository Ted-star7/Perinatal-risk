# 04 · Checklist baseline

The high-risk checklist is the baseline every model must beat. The flags are listed
in `contracts/cohort_config.yaml`:

- age under 18 or 35+, five or more previous live births, interval under 24 months,
  multiple pregnancy, previous stillbirth.

What to cover:
- build the checklist flag from those columns
- its flag rate, and how many perinatal deaths it catches
- this flag rate is what the model is compared against (same share flagged →
  which catches more events).
