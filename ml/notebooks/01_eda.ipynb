Goal: understand each cohort before modelling. KE squad runs Kenya, TZ squad runs Tanzania.

Clear all outputs before committing (keeps diffs clean, avoids leaking data).


```python
import pandas as pd

# KE squad: COUNTRY = 'ke'   |   TZ squad: COUNTRY = 'tz'
COUNTRY = 'ke'
COHORT  = 'perinatal'   # perinatal | perinatal_anc | lbw

df = pd.read_csv(f'../../data/processed/{COUNTRY}_{COHORT}.csv')
print(df.shape)
df.head()
```

## Shape, columns and dtypes
- row count matches the proposal (perinatal 19822/10980, anc 10590/5952, lbw 4299/4626)
- confirm dtypes; note the id/design columns `caseid`, `v001`, `v005`


```python
df.info()
df.describe(include='all').T
```

## Missingness
- which predictors have blanks? expect blanks in `told_high_bp`, `preceding_interval_m`
  (first pregnancy), and the ANC columns in the *_anc cohort


```python
df.isna().mean().sort_values(ascending=False)
```

## Outcome rate
- `perinatal_death` (perinatal / anc cohorts) or low birth weight from `birth_weight_g < 2500` (lbw)
- unweighted here; the published weighted rate is checked in notebook 03


```python
df['perinatal_death'].mean()   # swap to the lbw outcome for the lbw cohort
```

## Predictor distributions
Predictors: maternal_age, prior_pregnancies, prior_live_births, prior_stillbirth, prior_loss, preceding_interval_m, multiple, education, wealth_quintile, residence, region, told_high_bp.
- age, parity and interval distributions; categories for education / wealth / residence / region
- outcome rate within each category (early signal of what matters)


```python
# e.g. outcome rate by wealth quintile
df.groupby('wealth_quintile')['perinatal_death'].mean()
```
