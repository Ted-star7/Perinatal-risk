# 03 · Cohort provenance + the build gate

The three cohorts per country are already materialised in `data/processed/`.
This notebook records where they came from and runs the one gate that proves the
outcome is defined correctly.


```python
import pandas as pd

# KE squad: COUNTRY = 'ke'   |   TZ squad: COUNTRY = 'tz'
COUNTRY = 'ke'
COHORT  = 'perinatal'   # perinatal | perinatal_anc | lbw

df = pd.read_csv(f'../../data/processed/{COUNTRY}_{COHORT}.csv')
print(df.shape)
df.head()
```

## Row-count check
Confirm counts match contracts/cohort_config.yaml.


```python
expected = {'perinatal': {'ke':19822,'tz':10980},
            'perinatal_anc': {'ke':10590,'tz':5952},
            'lbw': {'ke':4299,'tz':4626}}
assert len(df) == expected[COHORT][COUNTRY], 'row count mismatch'
print('row count OK')
```

## Gate: weighted Kenya stillbirth rate ≈ 15.1 / 1000
- use `stillbirth` and the `v005` weight (÷1,000,000) on the Kenya perinatal cohort
- this is the proposal's proof the outcome is right; it must land near 15.1


```python
# KE perinatal cohort only
w = df['v005']/1_000_000
rate = (df['stillbirth']*w).sum()/w.sum()*1000
print(round(rate,1), 'per 1000')
```
