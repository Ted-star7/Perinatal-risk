# 05 · Models

Logistic regression, random forest, gradient boosting. Use grouped cross-validation
on the survey cluster (`v001`) so neighbouring households don't leak between train and
test. PR-AUC is the headline metric (events are ~3%).

What to cover:
- features (maternal_age, prior_pregnancies, prior_live_births, prior_stillbirth, prior_loss, preceding_interval_m, multiple, education, wealth_quintile, residence, region, told_high_bp) and the target `perinatal_death`
- class weights for the imbalance; grouped CV by `v001`
- PR-AUC and ROC-AUC per fold, with bootstrap intervals
- start on the headline 677-event cohort and keep the model simple.
